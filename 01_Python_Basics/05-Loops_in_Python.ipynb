## Types of Loops

A loop repeats a block of code many times. Python has two loop keywords:

- **for loop:** repeats a fixed number of times, or once for each item
- **while loop:** repeats as long as a condition is True

Common loop patterns:

- **while True:** runs forever until `break` stops it
- **nested loop:** a loop inside another loop

```python
#for loop structure
#range(start,stop) goes from start up to stop-1
#print hello 8 times, end=",\n" adds a comma and moves to the next line
for i in range(1,9):
    print('hello',end=",\n")
#range(start,stop,step) starts at 1, jumps by 3, stops before 6, so we get 1 and 4
for i in range(1,6,3):
    print(i)
```

```python
#range with step 2: start at 0, stop before 5, jump by 2
for i in range(0,5,2):
    print(i)
```

```python
#table of 7 using a for loop
n=7
#i goes from 1 to 10 (11 is not included)
for i in range(1,11):
#print one multiplication fact, for example 7 * 1 = 7
    print(n,"*",i,"=",n*i)
```

```python
#print numbers between 1 and 100 that are divisible by both 8 and 12
for i in range(1,101):
# % gives the remainder, and remainder 0 means exactly divisible
    if i%8==0 and i%12==0:
        print(i)
```

**Printing Star Pattern Using Loops**

```python
#star pattern: a triangle going up, then going down
#going up: row i prints i stars (1 to 5)
for i in range(1,6):
#inner loop prints one star each round, end="" keeps stars on the same line
    for j in range(i):
        print("*",end="")
    print()
#going down: rows print 4 stars, then 3, 2, 1
for i in range(4,0,-1):
    for j in range(i):
        print("*",end="")
    print()
```

```python
#while loop structure
#a while loop keeps running as long as the condition is True
n=2
#runs while n is not equal to 5
while n!=5:
    print(n)
#n must change inside the loop, otherwise the loop never ends
    n+=1
```

```python
#print a message 4 times using a while loop
i=1
while i<5:
    print("Hello Areej")
    i+=1
```

**Printing Table of 4**

```python
#table of 4 using a while loop
n=1
a=4
#n is the counter that goes from 1 to 10
while n<=10:
    print(a,'*',n,'=',n*a)
    n+=1
```

**Table of 5**

```python
#table of 5 using a while loop
n=5
i=1
while i<11:
    print(n,"*",i,"=",n*i)
    i+=1
```

```python
#while True is an infinite loop that stops only with break
#this program adds two numbers again and again until the user types yes
while True:
    num1=int(input("Enter a number here:"))
    num2=int(input("Enter second number here:"))
    print(num1+num2)
#ask the user if they want to stop
    repeat=input("Do you want to stop the program? (yes/no):")
    if repeat=="yes":
#break exits the loop immediately
        break
```

```python
#nested for loop: a loop inside another loop
#outer loop runs 5 times (5 rows)
for i in range(1,6):
#inner loop prints numbers 1 to 10 in every row
    for j in range(1,11):
#the last number gets a dot, all others get a comma
        if j==10:
            print(j,end=".")
        else:
            print(j,end=",")
    print()
```

**Loop Control Statements: Break and Continue**

```python
#break statement: stops the loop completely
for i in range(1,14):
    if i==7:
        break
    print(i)
#this line is outside the loop, so it prints once at the end
print("Thank you")
#continue statement: skips the current round and goes to the next one
for i in range(1,14):
    if i==7:
        continue
    print(i)
print("Thank you")
```

```python
#AI example 1: splitting a dataset into batches
#models are trained on small batches of data instead of all the data at once
total_samples=1000
batch_size=250
#range(start,stop,step) jumps by batch_size, just like range(1,6,3)
for start in range(0,total_samples,batch_size):
    end=start+batch_size
    print("Batch from sample",start,"to",end-1)
```

```python
#AI example 2: training loop with early stopping
#a model trains for many rounds (epochs), and we stop early if it stops improving
#validation loss after each epoch (lower is better)
val_losses=[0.90,0.72,0.61,0.55,0.54,0.56,0.57,0.58]
best_loss=100
#how many bad epochs in a row we allow
patience=2
bad_rounds=0
epoch=0
for loss in val_losses:
    epoch+=1
    if loss<best_loss:
#loss improved, so save it and reset the bad rounds counter
        best_loss=loss
        bad_rounds=0
        print("Epoch",epoch,",","loss",loss,",","improved")
    else:
#loss did not improve
        bad_rounds+=1
        print("Epoch",epoch,",","loss",loss,",","no improvement")
        if bad_rounds==patience:
            print("Early stopping at epoch:",epoch)
#break stops training early
            break
print("Best loss:",best_loss)
```

## What I Learned

- **for loop:** Repeats code for each value in a range.
- **range():** `range(stop)`, `range(start,stop)` and `range(start,stop,step)`, including counting backwards.
- **while loop:** Repeats while a condition is True, and the counter must change inside the loop.
- **while True:** An infinite loop that stops with `break`.
- **Nested loops:** Used a loop inside a loop to print star patterns and number rows.
- **break and continue:** `break` stops the loop, `continue` skips one round.
- **Loops with conditions:** Used `if` inside a loop to find numbers divisible by both 8 and 12.
- **AI example 1:** Split a dataset into batches with `range()` step.
- **AI example 2:** Built a training loop with early stopping using `for`, `if` and `break`.
