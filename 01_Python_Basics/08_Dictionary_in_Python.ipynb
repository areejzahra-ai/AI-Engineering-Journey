# Python Dictionaries

In this notebook, I learned how to store data as key:value pairs using dictionaries. I practiced creating a dictionary, accessing values by key, changing values, and adding and deleting keys. I also explored dictionary methods, looping, nested dictionaries, and finished with an AI engineering example.

```python
#dictionary basics: create, access, change, add and delete
#a dictionary stores data as key:value pairs inside curly brackets
book={"title":"Black Magic","author":"Areej Zahra","price":1500}
#access a value by its key
print(book["author"])
#change the value of an existing key
book["price"]=2000
print(book)
#add a new key called pages
book["pages"]=45
print(book)
#delete the author key
del book["author"]
print(book)
```

```python
#get(), in and len()
book={"title":"Black Magic","author":"Areej Zahra","price":1500}
#book["publisher"] would give a KeyError, but get() returns a default instead
print(book.get("author"))
print(book.get("publisher","Not available"))
#in checks if a key exists
print("price" in book)
print("pages" in book)
#len() counts the key:value pairs
print(len(book))
```

```python
#pop() removes a key and gives back its value
removed=book.pop("price")
print(removed)
print(book)
#update() adds or changes many keys at once
book.update({"price":1800,"pages":120})
print(book)
```

```python
#get() returns None (or a default) instead of an error when the key is missing
book={"title":"Black Magic","author":"Areej Zahra","price":1500}
print(book.get("author"))
print(book.get("publisher","Not available"))
#in checks if a key exists
print("price" in book)
```

```python
#nested dictionary: a dictionary inside a dictionary
student={"name":"Areej","marks":{"math":90,"python":95}}
print(student["marks"]["python"])
```

```python
#practice: driver data (access, change, add, delete, keys and values)
driver={"name":"Ahmed","city":"Karachi","rides":120,"rating":4.8,"earnings":25000}
#print the driver name
print(driver["name"])
#increase the earnings to 28000
driver["earnings"]=28000
#add a new key bonus with value 3000
driver["bonus"]=3000
#delete the city key
del driver["city"]
#print the final dictionary
print(driver)
#print only all the keys
print(driver.keys())
#print only all the values
print(driver.values())
```

```python
#practice: country sales (finding highest and lowest)
sales={"USA":120000,"UK":95000,"India":180000,"Canada":7000}
#find the country with the highest sales
highest_country="USA"
highest_sale=sales["USA"]
#items() gives each key and value together, so we unpack them
for country,sale in sales.items():
    if highest_sale<sale:
        highest_sale=sale
        highest_country=country
print(highest_sale,highest_country)
#find the country with the lowest sales
lowest_sale=sales["USA"]
lowest_country="USA"
for country,sale in sales.items():
    if lowest_sale>sale:
        lowest_sale=sale
        lowest_country=country
print(lowest_country,lowest_sale)
```

```python
#practice: employee data (nested dictionary)
#each employee ID is a key, and its value is another dictionary with details
employees={
    "E101":{"name":"Ali","dept":"Engineering","salary":85000},
    "E102":{"name":"Sara","dept":"Marketing","salary":65000},
    "E103":{"name":"Zain","dept":"Engineering","salary":95000},
    "E104":{"name":"Nida","dept":"Sales","salary":55000}
}
#print the names of all employees who work in engineering
#info is the inner dictionary of each employee
for employee_id,info in employees.items():
    if info["dept"]=="Engineering":
        print(info["name"])
#find the employee with the highest salary: start with the first employee
highest_salary=employees["E101"]["salary"]
highest_employee="Ali"
for employee_id,info in employees.items():
    #replace the saved values whenever a higher salary is found
    if highest_salary<info["salary"]:
        highest_salary=info["salary"]
        highest_employee=info["name"]
print(highest_salary,highest_employee)
#find the employee with the lowest salary
lowest_salary=employees["E101"]["salary"]
lowest_employee="Ali"
for employee_id,info in employees.items():
    if lowest_salary>info["salary"]:
        lowest_salary=info["salary"]
        lowest_employee=info["name"]
print(lowest_employee,lowest_salary)
```

```python
#practice: word frequency counter
#count how many times each word appears in a sentence (a common NLP task)
sentence="the quick brown fox jumps over the lazy dog the fox runs"
#split() breaks the sentence into a list of words
words=sentence.split()
print(words)
#word_count will store word:count pairs
word_count={}
for word in words:
    #check if the word is already a key in word_count (not in words)
    if word in word_count:
        #word seen before, so add 1 to its count
        word_count[word]+=1
    else:
        #first time we see this word, so start its count at 1
        word_count[word]=1
print(word_count)
```

```python
#AI example: a model configuration stored as a dictionary
#AI projects keep settings (hyperparameters) in a dictionary
config={"model":"Random Forest","learning_rate":0.01,"epochs":10,"batch_size":32}
#print every setting
for key,value in config.items():
    print(key,":",value)
#change one setting, then add a new one
config["epochs"]=20
config["optimizer"]="adam"
#get() safely reads a setting that may not exist
print(config.get("dropout",0.0))
print(config)
```
