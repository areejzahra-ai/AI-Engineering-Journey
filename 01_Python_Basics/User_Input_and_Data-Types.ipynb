## Python User Input and Data Types
In this notebook, I practiced taking input from the user and working with different data types in Python. I also built a small student identity card program and learned how to convert values using `int()`, `float()` and `str()`.

```python
#user input
#input() takes a value from the user and stores it in a variable
#whatever the user types is always stored as a string (str)
name=input("Enter your name here:")
print(name)
```

```python
#Student identity card information
#input() always returns a string, so we convert it with str(), int() and float()
name=str(input('Enter your name here'))
age=int(input('Enter your age here:'))
phone_number=input("Enter your number here:")
height=float(input("Enter your height here:"))
print(name)
print(age)
print(height,',',phone_number)                   
```

```python
#data types
#type() tells us the data type of a variable
#int = whole number, float = decimal number, str = text
name="areej"
age=45
height=6.2
print(type(height))
print(type(age))
print(type(name))
```

```python
#type conversion with user input
#input() always gives a string, so we check the type first and then convert it
a=input("Enter first number:")
b=input("Enter second number:")
print(type(a))
print(type(b))
print(a+b)

a=int(a)
b=int(b)
print(type(a))
print(type(b))
print(a+b)
```

## What I Learned

- **User input:** How to take input from the user with `input()` and store it in a variable.
- **Strings from input:** `input()` always returns a string, no matter what the user types.
- **Data types:** How to check a variable's type with `type()` (`int`, `float`, `str`).
- **Type conversion:** How to convert values using `str()`, `int()` and `float()`.
- **Type conversion with input:** Took two numbers from the user, checked that they are strings, and converted them with `int()` to add them correctly.
- **Practical example:** Built a student identity card program that takes name, age, phone number and height.
