# Practice: Loops and Number Problems

In this notebook, I practiced loops and conditions on number problems: f-strings, leap year, multiplication tables, counting, sums, factorial, even and odd sums, perfect numbers and prime numbers. I finished with an AI engineering example.

```python
#f-string: put variables inside text using {}
name="areej"
age=24
print(f"My name is {name}. My age is {age}.")
```

```python
#leap year: divisible by 400, or divisible by 4 but not by 100
year=int(input("Enter the year here:"))
if year%100==0 and year%400==0:
    print("leap year")
elif year%100!=0 and year%4==0:
    print("leap year")
else:
    print("not a leap year")
```

```python
#table of 5 using a for loop (starts at 1)
num=5
for i in range(1,11):
    print(num,"*",i,"=",num*i)
```

```python
#nested loop: the outer loop picks the number, the inner loop prints its table
for i in range(5,6):
    for j in range(1,11):
        print(i,"*",j,"=",i*j)
    #print() is outside the inner loop, so it adds one blank line after the table
    print()
```

```python
#table of a number given by the user
num=int(input("Enter the number here:"))
for i in range(1,11):
    print(num,"*",i,"=",num*i)
```

```python
#print Hello world n times
n=int(input("Enter your number here:"))
for i in range(n):
    print("Hello world")
```

```python
#print natural numbers from 1 to n
n=int(input("Enter the number here:"))
for i in range(1,n+1):
    print(i)
```

```python
#print n down to 1: step -1 counts backwards, and 0 is not included
n=int(input("Enter a number here:"))
for i in range(n,0,-1):
    print(i)
```

```python
#sum of the first n natural numbers
#total is used instead of sum, because sum is a built-in Python function
total=0
n=int(input("Enter a number here:"))
for i in range(1,n+1):
    total+=i
print(total)
```

```python
#factorial: multiply all numbers from 1 to n
f=1
n=int(input("Enter a number here:"))
for i in range(1,n+1):
    f=f*i
print(f)
```

```python
#sum of even numbers and sum of odd numbers from 1 to n
odd_sum=0
even_sum=0
n=int(input("Enter a number here:"))
for i in range(1,n+1):
    #remainder 0 means even
    if i%2==0:
        even_sum+=i
    else:
        odd_sum+=i
print("Sum of even numbers is:",even_sum)
print("Sum of odd numbers is:",odd_sum)
```

```python
#perfect number: the sum of its divisors (without the number itself) equals the number
#example: 28 = 1+2+4+7+14
divisor_sum=0
n=int(input("Enter your number here:"))
for i in range(1,n):
    if n%i==0:
        divisor_sum+=i
if divisor_sum==n:
    print("Perfect number")
else:
    print("Not a perfect number")
```

```python
#prime number: it has exactly 2 divisors, 1 and itself
count=0
n=int(input("Enter your number:"))
for i in range(1,n+1):
    if n%i==0:
        count+=1
if count==2:
    print("prime number")
else:
    print("Not a prime number")
```

```python
#AI example: split a dataset into batches and count the training steps
#models train on small batches, and one pass over all data is called an epoch
total_samples=1050
batch_size=100
batches=0
for start in range(0,total_samples,batch_size):
    batches+=1
print("Batches per epoch:",batches)
#training for 3 epochs
epochs=3
for epoch in range(1,epochs+1):
    print("Epoch",epoch,"finished after",batches,"batches")
print("Total training steps:",batches*epochs)
```
