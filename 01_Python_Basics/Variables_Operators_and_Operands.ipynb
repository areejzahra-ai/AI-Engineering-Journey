```python
#operators and oprends
#There are total seven types of operators
#arithematic operator
#logical operator
#assignment operator
#comparison operator
#membership operator
#identity operator
#Bitwise operator
#....................
#Arithematic operators
#there is seven type of arithematic operator
'''Addition+,subtraction_,multiplication*,division/,floor division//,** exponantion,% modulus'''
#floor division tells us about the decimal part 
a=3
b=5
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

```python
#comparison operators
#there are six type of comparison operators 
#greater then>,less than<,greater than equal to>=,less than equal to<=,equal to==,not equal to!=
a=6
b=7
print(a>b)
print(a<b)
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
```

```python
#assignment operators
#there are eight common types of assignment operators
#equal to =, add and assign +=, subtract and assign -=, multiply and assign *=,
#divide and assign /=, floor divide and assign //=, modulus and assign %=, power and assign **=
b=3
a=10
print(a)
a=10
a+=b
print(a)
a=10
a-=b
print(a)
a=10
a*=b
print(a)
a=10
a/=b
print(a)
a=10
a//=b
print(a)
a=10
a%=b
print(a)
a=10
a**=b
print(a) 
```

```python
#logical operators
#there are three logical operators
#and, or, not
a=6
b=7
print(a<b and b>5)
print(a>b and b>5)
print(a>b or b>5)
print(a>b or b<5)
print(not a<b)
print(not a>b)
```

```python
#membership operators
#there are two membership operators
#in, not in
a=[1,2,3,4,5]
b=3
c=9
print(b in a)
print(c in a)
print(b not in a)
print(c not in a)
```

```python
#identity operators
#there are two identity operators
#is, is not
a=[1,2]
b=a
c=[1,2]
print(a is b)
print(a is c)
print(a is not c)
print(a==c)
```

```python
#bitwise operators
#there are six bitwise operators
#AND &, OR |, XOR ^, NOT ~, left shift <<, right shift >>
a=6
b=3
print(a&b)
print(a|b)
print(a^b)
print(~a)
print(a<<1)
print(a>>1)
```
