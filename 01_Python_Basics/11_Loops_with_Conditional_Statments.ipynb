# Loops with Conditional Statements

In this notebook, I combined loops with `if`, `elif` and `else` to make decisions inside a loop. I practiced filtering, counting, and skipping values with `continue`, using lists and dictionaries. Most examples use AI engineering data like models, predictions and confidence scores.

```python
#loop with if else: check every model in a list
models=["model_A","model_B","model_C","model_D"]
for model in models:
    #or means only one of the two conditions must be True
    if model=="model_A" or model=="model_C":
        print("selected model")
    else:
        print("not selected")
```

```python
#classify every prediction by its confidence level
predictions=[0.92,0.65,0.87,0.45]
for prediction in predictions:
    if prediction>=0.90:
        print("High Confidence")
    elif prediction>=0.70:
        print("Medium Confidence")
    else:
        print("Low Confidence")
```

```python
#continue skips the current round when the condition is True
for i in range(1,6):
    if i==3:
        continue
    print(i)
```

```python
#a counter variable increases by 1 in every round
counter=0
for round_number in range(0,5):
    counter+=1
    print("test",counter)
```

```python
#count how many confidence scores are 0.80 or more
count=0
confidence_scores=[0.91,0.67,0.84,0.52,0.95]
for score in confidence_scores:
    if score>=0.80:
        count+=1
#print is outside the loop, so it runs once at the end
print(count)
```

```python
#print the models whose score is 80 or more
results={"Model A":85,"Model B":72,"Model C":91,"Model D":68}
#items() gives each key and value together
for key,value in results.items():
    if value>=80:
        print(key)
```

```python
#AI example: find the best model and the average score
best_model=""
best_score=0
total=0
for key,value in results.items():
    total+=value
    #replace the saved model whenever a higher score is found
    if value>best_score:
        best_score=value
        best_model=key
print("Best model:",best_model,best_score)
print("Average score:",total/len(results))
```

```python
#AI example: skip missing values and stop at a bad value
#None means the value is missing, and -1 means the data is corrupted
samples=[12,None,18,25,None,-1,30]
clean=[]
for value in samples:
    if value is None:
        #skip missing values
        continue
    if value<0:
        #stop when corrupted data is found
        print("Corrupted data found, stopping")
        break
    clean.append(value)
print(clean)
```

## What I Learned

- **if / else in loops:** Checked every item in a list and gave a different result for each.
- **if / elif / else in loops:** Classified predictions into High, Medium and Low confidence.
- **continue and break:** `continue` skips one round, `break` stops the loop.
- **Counters:** Counted items that match a condition with `count+=1`.
- **Dictionaries:** Looped with `items()` and used `if` to filter models by score.
- **AI examples:** Found the best model and cleaned a dataset by skipping missing values.
- **Key lesson:** Put `print(count)` outside the loop, otherwise it prints in every round.
