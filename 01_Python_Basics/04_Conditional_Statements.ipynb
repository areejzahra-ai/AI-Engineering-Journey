# Python Conditional Statements

In this notebook, I learned how Python makes decisions using `if`, `elif` and `else`. I also solved a practice question that combines comparison operators with logical operators (`and`, `or`).

```python
#if statement
#the code inside if runs only when the condition is True
age=20
if age>=18:
    print("You can vote")
```

```python
#if else statement
#else runs when the if condition is False
marks=40
if marks>=50:
    print("Pass")
else:
    print("Fail")
```

```python
#if elif else statement
#python checks conditions from top to bottom and runs only having first True one condition.
marks=85
if marks>=90:
    print("Grade A")
elif marks>=80:
    print("Grade B")
elif marks>=70:
    print("Grade C")
else:
    print("Grade D")
```

```python
#practice question: employee performance review
#this uses comparison operators (>=) and logical operators (and, or) together
#employee data (the values we will check)
#performance score out of 100
performance_score=78
#how many years the employee has worked here
years_at_company=4
#rating given by the manager out of 5
manager_rating=4.2
#True if the employee has a professional certification, otherwise False
has_certification=True
#number of projects the employee has finished
projects_completed=12
#condition 1: both parts must be True because of "and"
#score must be 90 or more AND manager rating must be 4.5 or more
if performance_score>=90 and manager_rating>=4.5:
    print("Exceptional")
#condition 2: runs only if condition 1 was False
#score must be 70 or more AND at least one of these must be True: 3+ years or certification
#the brackets make python check the "or" part first
elif performance_score>=70 and (years_at_company>=3 or has_certification):
    print("High performer")
#condition 3: runs only if conditions 1 and 2 were False
#score must be 70 or more AND projects completed must be 10 or more
elif performance_score>=70 and projects_completed>=10:
    print("Solid contributor")
#runs if none of the conditions above is True
else:
    print("Needs improvement")
```

```python
#practice question: should we deploy this machine learning model?
#in AI engineering, we check accuracy, overfitting and speed before deploying a model
train_accuracy=0.97
test_accuracy=0.88
latency_ms=120
#model must be accurate on new data (test) and fast enough, so both conditions must be True
if test_accuracy>=0.90 and latency_ms<=200:
    print("Deploy the model")
#a big gap between train and test accuracy means the model memorized the training data
elif train_accuracy-test_accuracy>0.05:
    print("Overfitting: add more data or use regularization")
#accuracy is not bad, so we can improve it by tuning hyperparameters
elif test_accuracy>=0.80:
    print("Decent model: tune the hyperparameters")
#runs if none of the above conditions is True
else:
    print("Poor model: retrain with better data")
```

```python
#practice question: routing a request to an LLM
#AI apps check the prompt before sending it to a model, to control cost and safety
prompt_tokens=3500
max_context=4096
is_sensitive=False
user_plan="free"
#the prompt is longer than the model can read
if prompt_tokens>max_context:
    print("Prompt too long: shorten it or split it")
#sensitive data must not be sent to the model
elif is_sensitive:
    print("Blocked: sensitive data found")
#premium users, or short prompts, can use the big model
elif user_plan=="premium" or prompt_tokens<=1000:
    print("Use the large model")
#everyone else uses the cheaper small model
else:
    print("Use the small model")
```

## What I Learned

- **if:** Runs a block of code only when a condition is True.
- **else:** Runs when the `if` condition is False.
- **elif:** Checks more conditions in order, and only the first True one runs.
- **Indentation:** Code inside `if`, `elif` and `else` must be indented.
- **Operators in conditions:** Used comparison operators (`>=`, `==`) and logical operators (`and`, `or`) to build real conditions.
