# Lists in Python

A **list** is an ordered, changeable collection of items written inside square brackets `[ ]` and separated by commas.

- **Ordered**: items keep their position
- **Indexed**: the first item is at index 0
- **Mutable**: items can be changed after the list is created
- **Allows duplicates**: the same value can appear more than once
- **Mixed types**: a list can hold numbers, strings, and other lists

```python
# Structure of a list
fruits = ["apple", "banana", "cherry", "banana"]

print(fruits[0])      # first item -> apple
print(fruits[-1])     # last item (negative index) -> banana
print(fruits[1:3])    # slicing: index 1 up to (not including) 3 -> ['banana', 'cherry']
print(len(fruits))    # number of items -> 4

fruits[0] = "mango"   # lists are mutable: change an item
print(fruits)         # ['mango', 'banana', 'cherry', 'banana']

mixed = [1, "hello", 3.5, True]   # different data types
nested = [[1, 2], [3, 4]]         # list inside a list
print(nested[1][0])               # row 1, item 0 -> 3
```

### How indexing works

| item  | apple | banana | cherry |
|-------|-------|--------|--------|
| index | 0     | 1      | 2      |
| negative index | -3 | -2 | -1   |

```python
# Method 1: built-in functions
c = [4, 8, 2, 9, 1]
greatest = max(c)
index = c.index(greatest)
print(f"the index of greatest number is {index}, the greatest number is {greatest}")

# Method 2: manual loop (same result, without max() and index())
c = [4, 8, 2, 9, 1]
greatest = c[0]     # assume the first element is the greatest
index = 0
for i in range(len(c)):
    if greatest < c[i]:   # found a bigger number, so update both values
        greatest = c[i]
        index = i
print(f"greatest number is {greatest}, at index {index}.")
```
