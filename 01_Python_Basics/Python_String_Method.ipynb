# Python String Methods

In this notebook, I explored built-in string methods in Python, such as `lower()`, `upper()`, `count()`, `find()` and `strip()`. I also built two practical examples: cleaning user input, and preparing text for an AI model (text preprocessing).

```python
#string methods
#a string is text inside quotes, and methods are built-in tools to work on it
a="Harry potter and goblet of fire"
#len() counts all characters, including spaces
print(len(a))
#lower() converts every letter to small letters
print(a.lower())
#count() tells how many times a word appear appears
print(a.count("t"))
#upper() converts every letter to capital letters
print(a.upper())
#capitalize() makes only the first letter of the whole string capital
print(a.capitalize())
#title() makes the first letter of every word capital
print(a.title())
#find() gives the index where "potter" starts 
print(a.find("potter"))
#index() finds "l" only between positions 15 and 25 (gives an error if not found)
print(a.index("l",15,25))
```

```python
#practical example: cleaning user input
#users often type extra spaces or mixed capital letters, so we clean the data first
name=input("Enter your full name:")
email=input("Enter your email:")
#strip() removes extra spaces from both ends, title() fixes the capital letters
clean_name=name.strip().title()
#lower() makes the email all small letters
clean_email=email.strip().lower()
print("Name:",clean_name)
print("Email:",clean_email)
#len() counts the characters in the cleaned name
print("Name length:",len(clean_name))
#count() checks that the email has exactly one @
print("Valid @ sign:",clean_email.count("@")==1)
```

```python
#AI example: text preprocessing for an NLP model
#AI models need clean text, so we remove noise before giving it to the model
review="  I LOVED this Movie!!! The acting was GREAT, but the ending was too Long...  "
#strip() removes extra spaces from both ends
text=review.strip()
#lower() makes all letters small, so "GREAT" and "great" are treated as the same word
text=text.lower()
#replace() removes punctuation we do not need
text=text.replace("!","")
text=text.replace(",","")
text=text.replace(".","")
print("Clean text:",text)
#split() breaks the sentence into a list of words (these are called tokens)
tokens=text.split()
print("Tokens:",tokens)
#len() counts how many tokens we have
print("Token count:",len(tokens))
#count() checks how many times the word "the" appears
print("Count of 'the':",text.count("the"))
#join() combines tokens back into one string
print("Joined:"," ".join(tokens))
```

## What I Learned Today

- **Basic string methods:** `len()`, `lower()`, `upper()`, `capitalize()` and `title()` to measure and change the case of text.
- **Searching:** `count()`, `find()` and `index()` to count and locate text inside a string.
- **Cleaning:** `strip()` to remove extra spaces and `replace()` to remove or change characters.
- **Splitting and joining:** `split()` to break a sentence into tokens and `join()` to combine them back.
- **Practical example:** Cleaned user input (name and email) so data is stored in a consistent format.
- **AI example:** Built a text preprocessing pipeline (clean, tokenize, count) which is the first step in NLP and LLM projects.
- **Key lesson:** Strings cannot be changed in place, so methods like `replace()` return a new string that must be saved with `text=text.replace(...)`.
