# Practice: Strings and While Loops

In this notebook, I practiced loops on strings and numbers: palindromes, counting letters, digits and special characters, extracting digits with a while loop, and input validation with `while True`. I also filtered a list and finished with an AI engineering example.

## Part 1: Strings

```python
#palindrome: a word that reads the same forward and backward
word="racecar"
reverse=""
for i in word:
    #adding each letter in front builds the reversed word
    reverse=i+reverse
if word==reverse:
    print("palindrome")
else:
    print("Not a palindrome")
```

```python
#ord() gives the number code of a character: A=65, Z=90, a=97, z=122, 0=48, 9=57
print(ord("a"))
```

```python
#count letters, digits and special characters in a string
a="P@#yn26atED"
letters=0
special=0
digits=0
for i in a:
    #capital letters are 65 to 90, small letters are 97 to 122
    if (ord(i)>=65 and ord(i)<=90) or (ord(i)>=97 and ord(i)<=122):
        letters+=1
    #digits are 48 to 57
    elif ord(i)>=48 and ord(i)<=57:
        digits+=1
    #everything else is a special character
    else:
        special+=1
print("Special characters:",special,", Digits:",digits,", Letters:",letters)
```

```python
#extract the digits of a number from right to left
#% 10 gives the last digit, // 10 removes the last digit
num=int(input("Enter a number here:"))
while num>0:
    print(num%10)
    num=num//10
#the loop stops when num becomes 0
print(num)
```

```python
#check if a number is a palindrome by reversing its digits
a=int(input("Enter a number here:"))
#keep a copy, because a is changed inside the loop
copy=a
rev=0
while a>0:
    #move rev one place left and add the last digit of a
    rev=rev*10+a%10
    a=a//10
if rev==copy:
    print("It's a palindrome")
else:
    print("It's not a palindrome")
```

## Part 3: Lists and Input Validation

```python
#classify every prediction by confidence
predictions=[0.82,0.45,0.91,0.67,0.76]
for prediction in predictions:
    if prediction>=0.70:
        print("High confidence")
    else:
        print("Low confidence")
```

```python
#build a new list with only the values greater than 10
data=[12,5,23,8,19,3]
new_list=[]
for i in data:
    if i>10:
        #append() adds an item to the end of a list
        new_list.append(i)
print(new_list)
```

```python
#while True keeps asking until the input is valid, then break stops the loop
while True:
    num=int(input("Enter a positive even number here:"))
    #the number must be positive AND even
    if num>0 and num%2==0:
        print(f"You entered a positive even number: {num}")
        break
    else:
        print("Sorry, select again")
```

```python
#AI example: keep only reliable predictions
#models often ignore predictions below a confidence threshold
threshold=0.80
reliable=[]
for prediction in predictions:
    if prediction>=threshold:
        reliable.append(prediction)
print("Reliable predictions:",reliable)
print("Kept",len(reliable),"out of",len(predictions))
```

## What I Learned

- **Palindromes:** Reversed a word with a loop, and a number with `%` and `//`.
- **ord():** Used character codes to count letters, digits and special characters.
- **while loops:** Extracted the digits of a number until it became 0.
- **Input validation:** Used `while True` with `break` to keep asking until the input is valid.
- **Filtering:** Built a new list with `append()` and an `if` condition.
- **AI example:** Kept only predictions above a confidence threshold.
