# Python Sets

In this notebook, I learned how to store unique values using sets. I practiced removing duplicates, checking membership, adding and removing items, and using set operations (intersection, difference, union and symmetric difference) to compare groups of customers. I finished with AI engineering examples.

```python
#set basics
#a set stores unique values in curly brackets, with no duplicates and no order
numbers={1,2,2,3,3,3}
print(numbers)
print(type(numbers))
#sets have no index, so numbers[0] would give a TypeError
#an empty set must use set(), because {} creates an empty dictionary
empty=set()
print(type(empty))
#sets can be looped like lists
for n in numbers:
    print(n)
```

```python
#practice: unique user IDs
user_ids=[101,102,105,101,109,110,105,112,109,115]
#set() removes the duplicates from the list
unique_user=set(user_ids)
#print all unique user ids
print(unique_user)
#print the total number of unique users
print(len(unique_user))
#check whether user id 110 exists in the set
print(110 in unique_user)
#check whether user id 120 exists in the set
print(120 in unique_user)
```

```python
#practice: add and remove items in a set
products={"Laptop","Mouse","Keyboard","Monitor"}
#add() adds one item
products.add("Webcam")
print(products)
#remove() deletes a specific item (error if it is missing)
products.remove("Mouse")
print(products)
#pop() removes a random item, because sets have no order
products.pop()
print(products)
#check whether printer exists
print("Printer" in products)
#print the final set
print(products)
```

```python
#practice: customers of two stores
store_A={1001,1002,1003,1005,1007,1010}
store_B={1003,1004,1005,1008,1010,1012}
#customers who shop at both stores (intersection)
print(store_A&store_B)
#customers who shop only at store A (difference)
print(store_A-store_B)
#customers who shop only at store B
print(store_B-store_A)
#all unique customers (union)
unique_customers=store_A|store_B
print(unique_customers)
#total number of unique customers
print(len(unique_customers))
#customers who shopped at exactly one store (symmetric difference)
print(store_A^store_B)
```

```python
#extra set methods
a={1,2,3}
b={1,2,3,4,5}
#update() adds many items at once
a.update([3,4])
print(a)
#discard() removes an item and gives no error if it is missing
a.discard(99)
#issubset() checks if all items of a are inside b
print({1,2}.issubset(b))
#isdisjoint() checks that two sets share nothing
print({7,8}.isdisjoint(b))
```

```python
#AI example 1: unique vocabulary of a text
#NLP models build a vocabulary of unique words from the tokens
sentence="the model learns the data and the model improves"
tokens=sentence.split()
vocabulary=set(tokens)
print(len(tokens),"tokens,",len(vocabulary),"unique words")

#AI example 2: check for data leakage
#train and test data must not share any samples
train_ids={1,2,3,4,5,6}
test_ids={5,7,8}
leaked=train_ids&test_ids
if leaked:
    print("Data leakage found:",leaked)
else:
    print("No leakage")
```

## What I Learned

- **Sets:** Store unique values, have no order and no indexing.
- **Removing duplicates:** Converted a list to a set with `set()` and counted unique items with `len()`.
- **Membership:** Used `in` to check if a value exists, which is fast on sets.
- **Changing sets:** `add()`, `remove()`, `discard()`, `pop()` and `update()`.
- **Set operations:** Intersection `&`, difference `-`, union `|` and symmetric difference `^`.
- **Comparing sets:** `issubset()` and `isdisjoint()`.
- **AI examples:** Built a vocabulary of unique words and checked for data leakage between train and test data.
- **Key lesson:** `{}` makes an empty dictionary, so an empty set needs `set()`.
