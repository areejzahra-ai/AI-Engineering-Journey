# Python Tuples

In this notebook, I learned how to create and use tuples, access items by index, unpack values into variables, and loop through tuples. I solved practice questions on employees, deliveries, movies and sales, and finished with an AI engineering example.

```python
#tuple basics
#a tuple stores many values in round brackets and cannot be changed after creation
colors=("red","green","blue","green")
print(type(colors))
#index 0 is the first item, -1 is the last item
print(colors[0])
print(colors[-1])
#slicing: from index 1 up to (not including) 3
print(colors[1:3])
print(len(colors))
#count() tells how many times a value appears, index() finds its position
print(colors.count("green"))
print(colors.index("blue"))
print("red" in colors)
#a tuple with one item needs a comma, otherwise it is just a string
single=("hello",)
print(type(single))
```

```python
#tuples are immutable: colors[0]="yellow" would give a TypeError
#to "change" a tuple, build a new one
new_colors=("yellow",)+colors[1:]
print(new_colors)
```

```python
employee = ("Ali", 27, "Data Scientist", 95000)
#print name
#Method1
print(employee[0])
#Method2
name,age,profession,salary=employee
print("Name:",name)
#print employee salary
print("Salary:",salary)
#second method
print(employee[3])
#print every element of tuple using a loop
for items in employee:
      print(items)
#print the type of each element using a loop
for item in employee:
     print(type(item))
```

```python
#Amazon delivery challenge
delivery=("ORD1023", "Karachi", "Delivered", 5)
#print only the city
order_num,city,status,items=delivery
print("City:",city)
#print only the delivery status
print("Delivery_status:",status)
#print every item using a loop
for item in delivery:
    print(item)
#print the index of every item
for index,items in enumerate(delivery):
    print(index,items)
```

```python
movies = (("Inception", 148),("Avatar", 162),("Interstellar", 169),("Joker",122))
#print every movie name
for movie,minutes in movies:
    print("name:",movie)
#print movie longer than 150 minutes
for movie,minutes in movies:
    if (minutes>150):
        print(movie)
#Find the longest movie without using max
longest_movie=movies[0][0]
longest_minutes=movies[0][1]
for movie,minutes in movies:
    if(longest_minutes<minutes):
        longest_minutes=minutes
        longest_movie=movie
print(longest_movie)
```

```python
employees=(("Sara", 28,4.7),("Ahmed",32,4.2),("John",26,4.9),("Fatima",30,4.5))
#print employee whos rating is greater than 4.5
for employee,age,rating in employees:
    if(rating>4.5):
       print(employee)
#Find the employee with the highest rating
highest_rating=employees[0][2]
highest_employee=employees[0][0]
for employee,age,rating in employees:
    if(highest_rating<rating):
        highest_rating=rating
        highest_employee=employee
print(highest_employee)
#calculate the average rating
total_rating=0
for employee,age,rating in employees:
    total_rating+=rating
average=total_rating/len(employees)
print(average)
```

```python
Sales= (("Monday", 120),("Tuesday",150),("Wednesday",90),("Thursday",180),("Friday",160))
#Find the day with the highest sales
highest_sale=Sales[0][1]
highest_day=Sales[0][0]
for day,sale in Sales:
    if(highest_sale<sale):
        highest_sale=sale
        highest_day=day
print(highest_sale,highest_day)
#find the day with lowest sales
lowest_sale=Sales[0][1]
lowest_day=Sales[0][0]
for day,sale in Sales:
    if(lowest_sale>sale):
        lowest_sale=sale
        lowest_day=day
print(lowest_sale,lowest_day)
#Calculate the average sale
total_sale=0
count=0
for day,sales in Sales:
        total_sale+=sale
        count+=1
average=total_sale/count
print(average)
#print each day whose sales are greater than average
for day,sales in Sales:
    if(sales>average):
        print(day)
       
```

```python
#AI example 1: dataset shape is a tuple
#NumPy and Pandas return the shape of data as a tuple (rows, columns)
dataset_shape=(1000,20)
rows,columns=dataset_shape
print("Samples:",rows,"Features:",columns)

#AI example 2: choose the best model that is fast enough
#each tuple is (model name, accuracy, latency in ms)
models=(("Logistic Regression",0.84,15),("Random Forest",0.91,120),("Neural Network",0.93,450),("KNN",0.88,60))
best_name=""
best_accuracy=0
for name,accuracy,latency in models:
    #skip models that are too slow, then keep the most accurate one
    if latency<=200 and accuracy>best_accuracy:
        best_accuracy=accuracy
        best_name=name
print("Best model:",best_name,best_accuracy)
```
