# Python Functions

In this notebook, I learned how to write reusable code with functions. I practiced functions with and without parameters, `return`, positional, default and keyword arguments, and returning multiple values. I finished with AI engineering examples.

```python
#functions: a function is a reusable block of code
#def creates the function, and calling its name runs the code inside
def greet():
    print("Hello!")
#call the function twice
greet()
greet()

def school():
    print("My school starts at 8 AM.")
school()

def favourite_food():
    print("pizza")
#a loop can call a function many times
for i in range(4):
    favourite_food()

def motivation():
    print("Practice makes a man perfect")
for i in range(5):
    motivation()
```

```python
#\n moves the text to a new line
def daily_routine():
    print("I wake up early in the morning.\nDid my breakfast.\nGo for office.\nStart my work.")
daily_routine()

def introduce():
    print("My name is Ali.\nI am learning Python.")
introduce()
introduce()

def favourite_subject():
    print("My favourite subject is Python.")
    print("I love to practice it daily.")
favourite_subject()

#a function can contain many lines of code
def morning_routine():
    print("I wake up early.")
    print("I brush my teeth.")
    print("I eat breakfast.")
    print("I go to school.")
for i in range(3):
    morning_routine()
```

```python
#parameters let a function work with different values
def greet_user(name):
    print("Hello,",name)
greet_user("Ali")
greet_user("Sara")
```

```python
#return sends a value back, so we can save it in a variable
def add(a,b):
    return a+b
result=add(5,3)
print(result)

#a function can use if and return different results
def check_even(n):
    if n%2==0:
        return "Even"
    else:
        return "Odd"
print(check_even(10))
print(check_even(7))
```

```python
#positional arguments: values go in the same order as the parameters
def multiplication(a,b,c,d):
    return a*b*c*d
result=multiplication(3,2,4,2)
print(result)

#default arguments: c and d use 3 and 5 if we do not give them
#this redefines multiplication with new parameters
def multiplication(a,b,c=3,d=5):
    return a*b*c*d
result=multiplication(3,8)
print(result)

#keyword arguments: we name each value, so the order does not matter
def multiplication(a,b,c,d):
    return a*b*c*d
result=multiplication(3,d=8,b=4,c=9)
print(result)
```

```python
#a function can return many values, and Python packs them into a tuple
def teacher():
    return "Areej",27,"Mathematics teacher"
info=teacher()
print(type(info))
#tuple unpacking: each value goes into its own variable
name,age,occupation=info
print(name)
print(age)
print(occupation)

#tuple methods: index() finds a position, count() counts a value
b=(4,"Areej",24,"Student",56.0,True,4,56.0,4,"Areej")
print(b.index("Student"))
print(b.count(4))
```

```python
#AI example 1: a reusable function to clean text
def clean_text(text):
    #strip() removes extra spaces and lower() makes all letters small
    text=text.strip().lower()
    #remove each punctuation mark one by one
    for mark in "!,.?":
        text=text.replace(mark,"")
    return text
print(clean_text("  I LOVED this Movie!!!  "))

#AI example 2: evaluate a model, with a default threshold and two return values
def evaluate_model(correct,total,threshold=0.9):
    accuracy=correct/total
    if accuracy>=threshold:
        status="Deploy"
    else:
        status="Retrain"
    return accuracy,status
acc,status=evaluate_model(88,100)
print(acc,status)
acc,status=evaluate_model(88,100,threshold=0.85)
print(acc,status)
```

## What I Learned

- **Functions:** Wrote reusable code with `def` and called it by name.
- **Parameters:** Passed values into functions.
- **return:** Sent a value back and saved it in a variable (`return` is different from `print`).
- **Arguments:** Used positional, default and keyword arguments.
- **Multiple return values:** Returned a tuple and unpacked it into variables.
- **Functions with loops and conditions:** Called functions inside loops and used `if` inside functions.
- **AI examples:** Built a text cleaning function and a model evaluation function.
- **Key lesson:** A function runs only when it is called, and defining it again with the same name replaces the old one.
